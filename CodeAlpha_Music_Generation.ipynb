```python
import pretty_midi

# Create a new MIDI file
midi = pretty_midi.PrettyMIDI()

# Create a piano instrument
piano = pretty_midi.Instrument(program=0)

# Musical notes for a sample melody
notes = [60, 62, 64, 65, 67, 65, 64, 62, 60]

for i, pitch in enumerate(notes):
    note = pretty_midi.Note(
        velocity=100,
        pitch=pitch,
        start=i * 0.5,
        end=(i + 1) * 0.5
    )
    piano.notes.append(note)

midi.instruments.append(piano)

# Save the sample melody
midi.write("sample_melody.mid")

print("Sample musical data created successfully!")
```

### Generating a New Melody with a Faster Tempo

```python
import pretty_midi
from IPython.display import Audio, display

# Create a new MIDI object with a faster tempo
new_midi = pretty_midi.PrettyMIDI(initial_tempo=140)

# Let's use a bright instrument, like a Marimba (program 12) or Synth Brass (program 62)
# Let's go with Marimba (12)
instrument = pretty_midi.Instrument(program=12)

# Define notes (pitch, duration, and rhythm)
# A rhythmic pattern in G major
notes_sequence = [
    (55, 0.0, 0.2),   # G3
    (59, 0.2, 0.4),   # B3
    (62, 0.4, 0.6),   # D4
    (67, 0.6, 1.0),   # G4
    (62, 1.0, 1.2),   # D4
    (59, 1.2, 1.4),   # B3

    (57, 1.4, 1.6),   # A3
    (61, 1.6, 1.8),   # C#4
    (64, 1.8, 2.0),   # E4
    (69, 2.0, 2.4),   # A4
    (64, 2.4, 2.6),   # E4
    (61, 2.6, 2.8),   # C#4

    (59, 2.8, 3.0),   # B3
    (62, 3.0, 3.2),   # D4
    (66, 3.2, 3.4),   # F#4
    (71, 3.4, 3.8),   # B4
    (74, 3.8, 4.2),   # D5
    (71, 4.2, 4.6),   # B4
    (67, 4.6, 5.0),   # G4
]

for pitch, start, end in notes_sequence:
    note = pretty_midi.Note(
        velocity=105,
        pitch=pitch,
        start=start,
        end=end
    )
    instrument.notes.append(note)

new_midi.instruments.append(instrument)

# Save the new melody
new_midi.write("fast_melody.mid")

# Convert to audio and play
new_audio = new_midi.fluidsynth()
display(Audio(new_audio, rate=44100))
```
