```python

# CodeAlpha Internship - Task 2
# FAQ Chatbot

faqs = {
    "what is your name": "I am an FAQ Chatbot.",
    "what is python": "Python is a popular programming language.",
    "what is artificial intelligence": "Artificial Intelligence enables machines to perform tasks that require human intelligence.",
    "what is machine learning": "Machine Learning is a branch of AI that helps computers learn from data.",
    "how can i contact support": "You can contact support through the official support email.",
    "what are your working hours": "Our working hours are 9 AM to 5 PM.",
    "how do i register": "You can register by filling out the registration form.",
    "thank you": "You're welcome! Happy to help.",
    "bye": "Goodbye! Have a great day."
}

print("FAQ dataset loaded successfully!")
print("Total FAQs:", len(faqs))
```

```python

import re
from difflib import get_close_matches

def chatbot_response(user_input):
    # Convert input to lowercase and remove extra symbols
    question = re.sub(r'[^a-z0-9\s]', '', user_input.lower())
    question = " ".join(question.split())

    # Check for an exact match
    if question in faqs:
        return faqs[question]

    # Find a similar question
    matches = get_close_matches(question, faqs.keys(), n=1, cutoff=0.6)

    if matches:
        return faqs[matches[0]]

    return "Sorry, I don't know the answer. Please try another FAQ."

print("Chatbot logic created successfully!")
```

```python

print("Chatbot:", chatbot_response("What is Python?"))
print("Chatbot:", chatbot_response("What is AI?"))
print("Chatbot:", chatbot_response("Thank you"))
print("Chatbot:", chatbot_response("Tell me about cricket"))
```

```python

print("================================")
print("       FAQ CHATBOT")
print("================================")
print("Ask me a question! Type 'exit' to stop.\n")

while True:
    user_question = input("You: ")

    if user_question.lower().strip() == "exit":
        print("Chatbot: Thank you for chatting. Goodbye!")
        break

    answer = chatbot_response(user_question)
    print("Chatbot:", answer)
```

```python

# CodeAlpha Internship - Task 2
# FAQ Chatbot using Python and Gradio

!pip -q install gradio

import re
from difflib import get_close_matches
import gradio as gr

# Step 1: FAQ Dataset
faqs = {
    "what is your name": "I am the CodeAlpha FAQ Chatbot.",
    "what is python": "Python is a popular, high-level programming language.",
    "what is artificial intelligence": "Artificial Intelligence enables machines to perform tasks that normally require human intelligence.",
    "what is machine learning": "Machine Learning is a branch of AI that enables computers to learn from data.",
    "what is deep learning": "Deep Learning uses neural networks with multiple layers to learn complex patterns.",
    "what is data science": "Data Science uses data analysis and statistics to extract useful insights.",
    "how can i contact support": "You can contact support through the official support email.",
    "what are your working hours": "Our working hours are 9 AM to 5 PM.",
    "how do i register": "You can register by completing the registration form.",
    "what is your purpose": "I answer frequently asked questions quickly and simply.",
    "thank you": "You're welcome! Happy to help.",
    "bye": "Goodbye! Have a great day."
}

# Step 2: Process and match user questions
def chatbot_response(user_input, history=None):
    question = re.sub(r"[^a-z0-9\s]", "", user_input.lower())
    question = " ".join(question.split())

    # Exact match
    if question in faqs:
        return faqs[question]

    # Similarity match
    matches = get_close_matches(
        question, faqs.keys(), n=1, cutoff=0.65
    )

    if matches:
        return faqs[matches[0]]

    return (
        "Sorry, I couldn't find an answer to that question. "
        "Please try asking about Python, AI, machine learning, or support."
    )

# Step 3: Create the chat interface
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 CodeAlpha FAQ Chatbot",
    description=(
        "Welcome! Ask me questions about Python, AI, "
        "machine learning, registration, and support."
    ),
    examples=[
        "What is Python?",
        "What is artificial intelligence?",
        "What is machine learning?",
        "How do I register?",
        "What are your working hours?"
    ]
)

# Step 4: Launch the chatbot
demo.launch(share=True)
```
